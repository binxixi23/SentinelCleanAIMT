```python
# ==========================================
# STEP 1: INITIALIZE ENVIRONMENT DEPENDENCIES
# ==========================================
import os
import sys

# Automatically install core frameworks if executed inside a Google Colab instance
if 'google.colab' in sys.modules or os.path.exists('/content'):
    print("[Environment] Google Colab detected. Installing Ultralytics ecosystem...")
    os.system('pip install -q ultralytics')

import yaml
import cv2
import torch
import numpy as np
from ultralytics import YOLO

# Suppress external visualization ecosystems to prevent runtime parsing conflicts
os.environ["WANDB_DISABLED"] = "true"
os.environ["CLEARML_LOGGING"] = "false"

# Configure explicit absolute paths to comply with standard YOLO directory tree structure
BASE_DIR = os.getcwd()
DATA_DIR = os.path.join(BASE_DIR, "data")
IMAGES_DIR = os.path.join(DATA_DIR, "images")
LABELS_DIR = os.path.join(DATA_DIR, "labels")

# Create separate top-level tracks for images and bounding box coordinate txt files
os.makedirs(IMAGES_DIR, exist_ok=True)
os.makedirs(LABELS_DIR, exist_ok=True)

print(f"[System Status] Base working directory locked to: {BASE_DIR}")
print(f"[System Status] Training assets structured inside: {DATA_DIR}")

# ==========================================
# STEP 2: PROGRAMMATIC SYNTHETIC DATA GENERATION
# ==========================================
print("[Data Engineering] Generating mathematical hazard validation matrices via OpenCV...")

for frame_id in range(1, 3):
    mock_floor = np.ones((640, 640, 3), dtype=np.uint8) * 120

    if frame_id == 1:
        # Frame 1: Simulate custom fluid spill boundaries (Blue circle matrix)
        cv2.circle(mock_floor, (320, 350), 120, (255, 0, 0), -1)
        cv2.imwrite(os.path.join(IMAGES_DIR, "synthetic_sample_1.jpg"), mock_floor)

        # Isolate text annotations directly inside the standard /labels directory
        with open(os.path.join(LABELS_DIR, "synthetic_sample_1.txt"), "w") as f:
            f.write("1 0.5 0.5 0.3 0.3\n")

    else:
        # Frame 2: Simulate custom dry debris obstacles (White bounding box matrix)
        cv2.rectangle(mock_floor, (200, 200), (350, 350), (255, 255, 255), -1)
        cv2.imwrite(os.path.join(IMAGES_DIR, "synthetic_sample_2.jpg"), mock_floor)

        # Isolate text annotations directly inside the standard /labels directory
        with open(os.path.join(LABELS_DIR, "synthetic_sample_2.txt"), "w") as f:
            f.write("0 0.43 0.43 0.23 0.23\n")

print(f"[Data Engineering] Custom images and coordinate labels successfully separated and built.")

# ==========================================
# STEP 3: YAML DATA PLAN MAPPING CONFIGURATION
# ==========================================
dataset_schema = {
    'path': DATA_DIR,
    'train': 'images',
    'val': 'images',
    'names': {
        0: 'dry_debris',
        1: 'liquid_spill',
        2: 'charging_dock',
        3: 'pet_obstacle'
    }
}

YAML_PATH = os.path.join(DATA_DIR, "data_plan.yaml")
with open(YAML_PATH, 'w') as f:
    yaml.dump(dataset_schema, f)

print(f"[Configuration Plan] Structured YAML dataset schematic generated at: {YAML_PATH}")

# ==========================================
# STEP 4: NEURAL LAYER TRANSFER LEARNING EXECUTION
# ==========================================
print("[Model Execution] Launching custom YOLOv11 fine-tuning pipeline sequence...")

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"[Compute Hardware] Target execution runtime processor engine: {device.upper()}")

# Instantiate foundational pre-trained weights profile from workspace
WEIGHTS_PATH = os.path.join(BASE_DIR, "yolo11n.pt")
model = YOLO(WEIGHTS_PATH)

# Run Transfer Learning epoch pass with plotting engines bypassed to prevent conflicts
results = model.train(
    data=YAML_PATH,
    epochs=1,
    imgsz=640,
    device=device,
    plots=False,
    save=False,
    verbose=True
)

print("[Framework Success] Deployment logs and custom training parameters successfully locked into workspace directory.")
```
